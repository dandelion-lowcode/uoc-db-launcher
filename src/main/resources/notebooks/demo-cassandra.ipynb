# Cassandra quick start

Connect to the local Cassandra service and inspect the default session.

```python
# Install the Cassandra driver if needed
%pip install cassandra-driver
```

```python
# Import the Cassandra client
from cassandra.cluster import Cluster
```

```python
# Connect and check the Cassandra version
cluster = Cluster(['host.docker.internal'])
session = cluster.connect()
version = session.execute('SELECT release_version FROM system.local').one()
print(f'Cassandra version: {version.release_version}')
```

```python
# Close the connection
cluster.shutdown()
```
