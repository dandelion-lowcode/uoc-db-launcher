# MongoDB quick start

Connect to MongoDB, list databases, and inspect the `mensajeria` database.

```python
# Install the MongoDB driver if needed
%pip install pymongo
```

```python
# Import the MongoDB client
from pymongo import MongoClient
```

```python
# Connect to MongoDB
client = MongoClient('mongodb://host.docker.internal:27017/')
client.admin.command('ping')
```

```python
# List the available databases
client.list_database_names()
```

```python
# Select the mensajeria database
mensajeria_db = client['mensajeria']
mensajeria_db.list_collection_names()
```

```python
# Close the connection
client.close()
```
