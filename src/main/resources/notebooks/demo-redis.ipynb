# Redis quick start

Connect to Redis, write one value, and read it back.

```python
# Install the Redis driver if needed
%pip install redis
```

```python
# Import the Redis client
import redis
```

```python
# Connect and check Redis
r = redis.Redis(host='localhost', port=6379, db=0, decode_responses=True)
r.ping()
```

```python
# Set a value
r.set('k', 'hi')
```

```python
# Get a value
r.get('k')
```

```python
# Clean up
r.delete('k')
```
