# Neo4j (Twitter) quick start

Explore the imported Twitter graph through the launcher database service.

This service runs Neo4j 3.5, the version the dataset was built with. The same
driver talks to it and to the newer Neo4j beside it.


```python
# The driver Neo4j publishes, if it is not installed yet
%pip install neo4j
```

```python
# py2neo is no longer maintained. This is the driver Neo4j publishes and supports,
# and it speaks Bolt rather than HTTP, which is why the port is not the browser one.
from neo4j import GraphDatabase

driver = GraphDatabase.driver('bolt://host.docker.internal:17687')
driver.verify_connectivity()
```

```python
# execute_query returns the records, a summary, and the keys. Only the records are
# needed here, so the other two are thrown away.
def query(cypher):
    records, _, _ = driver.execute_query(cypher)
    return records

def one(cypher):
    return query(cypher)[0].values()[0]
```

```python
# Count the nodes
one('MATCH (n) RETURN count(n)')
```

```python
# Count the relationships
one('MATCH ()-[r]->() RETURN count(r)')
```

```python
# The labels in use
[r['label'] for r in query('CALL db.labels() YIELD label RETURN label ORDER BY label')]
```

```python
# A few tweets
for r in query('MATCH (t:Tweet) RETURN t.text AS text, t.createdAt AS created_at LIMIT 5'):
    print(r['created_at'], '|', r['text'])
```

```python
# Replies and retweets, and what they point at
for r in query('''MATCH (t:Tweet)-[link:IS_A_RETWEET_OF|IS_A_REPLY_OF]->(original:Tweet)
                  RETURN type(link) AS kind, t.text AS text LIMIT 5'''):
    print(r['kind'], '|', r['text'])
```

```python
# Closing the driver releases the connection to the database
driver.close()
```
